```python
from aqua.diagnostics import GlobalBiases
```

```python
catalog = 'climatedt-o25.1'
exp = 'historical-1990'
source = 'lra-r100-monthly'
loglevel = 'INFO'
startdate = '1990-01-01'
enddate = '2015-12-31'
var = 'u'
plev = 850

icon_dict = {
    'catalog': catalog,
    'model': 'ICON',
    'exp': exp,
    'source': source,
    'loglevel': loglevel,
    'startdate': startdate,
    'enddate': enddate,
    'var': var,
    'plev': plev
}

ifs_nemo_dict = {
    'catalog': catalog,
    'model': 'IFS-NEMO',
    'exp': exp,
    'source': source,
    'loglevel': loglevel,
    'startdate': startdate,
    'enddate': enddate,
    'var': var,
    'plev': plev
}

ifs_fesom_dict = {
    'catalog': catalog,
    'model': 'IFS-FESOM',
    'exp': exp,
    'source': source,
    'loglevel': loglevel,
    'startdate': startdate,
    'enddate': enddate,
    'var': var,
    'plev': plev
}
```

```python
gb_icon = GlobalBiases(**icon_dict)
gb_ifs_nemo = GlobalBiases(**ifs_nemo_dict)
gb_ifs_fesom = GlobalBiases(**ifs_fesom_dict)
```

```python
gb_icon.retrieve()
gb_icon.compute_climatology(seasonal=False)
```

```python
gb_ifs_fesom.retrieve()
gb_ifs_fesom.compute_climatology(seasonal=False)
```

```python
gb_ifs_nemo.retrieve()
gb_ifs_nemo.compute_climatology(seasonal=False)
```
